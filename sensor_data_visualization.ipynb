# Astro Pi Sensor Data Visualization

This notebook visualizes sensor data from the ISS Astro Pi mission. We explore:
- **Environmental**: Temperature, Humidity, Pressure
- **Magnetic**: MagX, MagY, MagZ, Compass
- **Orientation**: Pitch, Roll, Yaw
- **Acceleration**: AccelX, AccelY, AccelZ
- **Light/Color**: R, G, B, C (camera)
- **Position**: Latitude, Longitude
- **Motion**: Motion detection

```python
# Imports
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
import warnings
warnings.filterwarnings('ignore')

plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette("husl")
pd.set_option('display.max_columns', None)

df = pd.read_csv('AstroPI_CSV.csv')
df['Date/Time'] = pd.to_datetime(df['Date/Time'])

print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
df.head(10)
```

```python
# Plot 1: Environmental sensors over time (Temperature, Humidity, Pressure)
fig, axes = plt.subplots(3, 1, figsize=(14, 10), sharex=True)

env_cols = ['Temperature', 'Humidity', 'Pressure']
titles = ['Temperature (°C)', 'Humidity (%)', 'Pressure (hPa)']

for i, (col, title) in enumerate(zip(env_cols, titles)):
    axes[i].plot(df['Date/Time'], df[col], linewidth=0.6, alpha=0.8)
    axes[i].set_ylabel(title, fontsize=11)
    axes[i].set_title(title, fontsize=12, fontweight='bold')
    axes[i].grid(True, alpha=0.3)

axes[-1].set_xlabel('Time', fontsize=11)
plt.suptitle('Environmental Sensors Over Time', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
# Plot 2: Magnetic field components and compass over time
fig, axes = plt.subplots(4, 1, figsize=(14, 12), sharex=True)

mag_cols = ['MagX', 'MagY', 'MagZ', 'Compass']
for i, col in enumerate(mag_cols):
    axes[i].plot(df['Date/Time'], df[col], linewidth=0.6, alpha=0.8)
    axes[i].set_ylabel(col, fontsize=11)
    axes[i].set_title(col, fontsize=12, fontweight='bold')
    axes[i].grid(True, alpha=0.3)

axes[-1].set_xlabel('Time', fontsize=11)
plt.suptitle('Magnetic Field and Compass Over Time', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
# Plot 3: Orientation (Pitch, Roll, Yaw) over time
fig, axes = plt.subplots(3, 1, figsize=(14, 9), sharex=True)

orient_cols = ['Pitch', 'Roll', 'Yaw']
for i, col in enumerate(orient_cols):
    axes[i].plot(df['Date/Time'], df[col], linewidth=0.6, alpha=0.8)
    axes[i].set_ylabel(col, fontsize=11)
    axes[i].set_title(col, fontsize=12, fontweight='bold')
    axes[i].grid(True, alpha=0.3)

axes[-1].set_xlabel('Time', fontsize=11)
plt.suptitle('Orientation (Pitch, Roll, Yaw) Over Time', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
# Plot 4: Acceleration and motion over time
fig, axes = plt.subplots(4, 1, figsize=(14, 10), sharex=True)

accel_cols = ['AccelX', 'AccelY', 'AccelZ', 'Motion']
for i, col in enumerate(accel_cols):
    axes[i].plot(df['Date/Time'], df[col], linewidth=0.6, alpha=0.8)
    axes[i].set_ylabel(col, fontsize=11)
    axes[i].set_title(col, fontsize=12, fontweight='bold')
    axes[i].grid(True, alpha=0.3)

axes[-1].set_xlabel('Time', fontsize=11)
plt.suptitle('Acceleration and Motion Over Time', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
# Plot 5: Light/Color (R, G, B, C) over time
fig, axes = plt.subplots(4, 1, figsize=(14, 10), sharex=True)

color_cols = ['R', 'G', 'B', 'C']
for i, col in enumerate(color_cols):
    axes[i].plot(df['Date/Time'], df[col], linewidth=0.6, alpha=0.8)
    axes[i].set_ylabel(col, fontsize=11)
    axes[i].set_title(f'Color channel: {col}', fontsize=12, fontweight='bold')
    axes[i].grid(True, alpha=0.3)

axes[-1].set_xlabel('Time', fontsize=11)
plt.suptitle('Light/Color Channels Over Time', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
# Plot 6: ISS trajectory (Latitude, Longitude) over time
fig, axes = plt.subplots(2, 1, figsize=(14, 8), sharex=True)

axes[0].plot(df['Date/Time'], df['Latitude'], linewidth=0.6, alpha=0.8, color='blue')
axes[0].set_ylabel('Latitude', fontsize=11)
axes[0].set_title('Latitude Over Time', fontsize=12, fontweight='bold')
axes[0].grid(True, alpha=0.3)

axes[1].plot(df['Date/Time'], df['Longitude'], linewidth=0.6, alpha=0.8, color='green')
axes[1].set_ylabel('Longitude', fontsize=11)
axes[1].set_title('Longitude Over Time', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Time', fontsize=11)
axes[1].grid(True, alpha=0.3)

plt.suptitle('ISS Position Over Time', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
# Plot 7: Geographic path colored by Temperature
fig, ax = plt.subplots(figsize=(14, 8))
sc = ax.scatter(df['Longitude'], df['Latitude'], c=df['Temperature'], cmap='coolwarm', s=2, alpha=0.7)
ax.set_xlabel('Longitude', fontsize=12)
ax.set_ylabel('Latitude', fontsize=12)
ax.set_title('ISS Trajectory Colored by Temperature', fontsize=14, fontweight='bold')
plt.colorbar(sc, label='Temperature (°C)')
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Plot 8: Geographic path colored by Pressure and Humidity
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

sc0 = axes[0].scatter(df['Longitude'], df['Latitude'], c=df['Pressure'], cmap='viridis', s=2, alpha=0.7)
axes[0].set_title('Trajectory by Pressure', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Longitude'); axes[0].set_ylabel('Latitude')
plt.colorbar(sc0, ax=axes[0], label='Pressure (hPa)')

sc1 = axes[1].scatter(df['Longitude'], df['Latitude'], c=df['Humidity'], cmap='Blues', s=2, alpha=0.7)
axes[1].set_title('Trajectory by Humidity', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Longitude'); axes[1].set_ylabel('Latitude')
plt.colorbar(sc1, ax=axes[1], label='Humidity (%)')

plt.tight_layout()
plt.show()
```

```python
# Plot 9: Distribution histograms for all numeric sensors
sensor_cols = ['Temperature', 'Humidity', 'Pressure', 'Compass', 'MagX', 'MagY', 'MagZ',
               'Pitch', 'Roll', 'Yaw', 'AccelX', 'AccelY', 'AccelZ', 'R', 'G', 'B', 'C', 'Motion']

fig, axes = plt.subplots(4, 5, figsize=(20, 16))
axes = axes.flatten()

for i, col in enumerate(sensor_cols):
    ax = axes[i]
    ax.hist(df[col].dropna(), bins=50, edgecolor='black', alpha=0.7)
    ax.set_title(col, fontsize=10, fontweight='bold')
    ax.set_ylabel('Count')
    ax.grid(True, alpha=0.3)

for j in range(len(sensor_cols), len(axes)):
    axes[j].axis('off')

plt.suptitle('Distribution of All Sensor Values', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
# Plot 10: Correlation heatmap of all sensors
numeric_df = df.select_dtypes(include=[np.number])
corr = numeric_df.corr()

fig, ax = plt.subplots(figsize=(16, 14))
sns.heatmap(corr, annot=False, cmap='coolwarm', center=0, square=True, linewidths=0.5, ax=ax)
ax.set_title('Sensor Correlation Matrix', fontsize=14, fontweight='bold')
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Plot 11: Environmental sensors - scatter matrix (Temperature vs Humidity vs Pressure)
env = df[['Temperature', 'Humidity', 'Pressure']]
g = sns.pairplot(env, diag_kind='kde', plot_kws={'alpha': 0.3, 's': 15})
g.fig.suptitle('Environmental Sensors Pair Plot', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
# Plot 12: Magnetic field 3D components (MagX, MagY, MagZ) scatter
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

pairs = [('MagX', 'MagY'), ('MagX', 'MagZ'), ('MagY', 'MagZ')]
for ax, (x, y) in zip(axes, pairs):
    ax.scatter(df[x], df[y], alpha=0.2, s=5)
    ax.set_xlabel(x); ax.set_ylabel(y)
    ax.set_title(f'{x} vs {y}')
    ax.grid(True, alpha=0.3)

plt.suptitle('Magnetic Field Component Pairs', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
# Plot 13: Box plots for environmental, magnetic, orientation, acceleration, color
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

df[['Temperature', 'Humidity', 'Pressure']].boxplot(ax=axes[0, 0])
axes[0, 0].set_title('Environmental', fontsize=11, fontweight='bold')
axes[0, 0].tick_params(axis='x', rotation=45)

df[['Compass', 'MagX', 'MagY', 'MagZ']].boxplot(ax=axes[0, 1])
axes[0, 1].set_title('Magnetic', fontsize=11, fontweight='bold')
axes[0, 1].tick_params(axis='x', rotation=45)

df[['Pitch', 'Roll', 'Yaw']].boxplot(ax=axes[1, 0])
axes[1, 0].set_title('Orientation', fontsize=11, fontweight='bold')
axes[1, 0].tick_params(axis='x', rotation=45)

df[['AccelX', 'AccelY', 'AccelZ']].boxplot(ax=axes[1, 1])
axes[1, 1].set_title('Acceleration', fontsize=11, fontweight='bold')
axes[1, 1].tick_params(axis='x', rotation=45)

plt.suptitle('Box Plots of Sensor Values', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

fig, ax = plt.subplots(figsize=(10, 5))
df[['R', 'G', 'B', 'C']].boxplot(ax=ax)
ax.set_title('Color Channels', fontsize=12, fontweight='bold')
ax.tick_params(axis='x', rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Plot 14: Rolling mean of key sensors (smoothing)
window = 200
fig, axes = plt.subplots(3, 1, figsize=(14, 10), sharex=True)

for idx, col in enumerate(['Temperature', 'Pressure', 'Compass']):
    axes[idx].plot(df['Date/Time'], df[col], alpha=0.3, linewidth=0.5, label='Raw')
    axes[idx].plot(df['Date/Time'], df[col].rolling(window).mean(), linewidth=1.5, label=f'Rolling mean (n={window})')
    axes[idx].set_ylabel(col)
    axes[idx].set_title(f'{col} - Raw vs Rolling Mean', fontsize=12, fontweight='bold')
    axes[idx].legend()
    axes[idx].grid(True, alpha=0.3)

axes[-1].set_xlabel('Time')
plt.suptitle('Rolling Mean of Selected Sensors', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
# Plot 15: KDE (density) for Temperature, Humidity, Pressure
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
for ax, col in zip(axes, ['Temperature', 'Humidity', 'Pressure']):
    df[col].plot(kind='kde', ax=ax)
    ax.set_title(col, fontsize=12, fontweight='bold')
    ax.set_xlabel(col)
    ax.grid(True, alpha=0.3)
plt.suptitle('Density of Environmental Sensors', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
# Plot 16: Trajectory colored by Compass and by Mag magnitude
df['Mag_Magnitude'] = np.sqrt(df['MagX']**2 + df['MagY']**2 + df['MagZ']**2)

fig, axes = plt.subplots(1, 2, figsize=(16, 6))
sc0 = axes[0].scatter(df['Longitude'], df['Latitude'], c=df['Compass'], cmap='hsv', s=2, alpha=0.7)
axes[0].set_title('Trajectory by Compass (heading)', fontsize=12, fontweight='bold')
plt.colorbar(sc0, ax=axes[0], label='Compass (°)')

sc1 = axes[1].scatter(df['Longitude'], df['Latitude'], c=df['Mag_Magnitude'], cmap='plasma', s=2, alpha=0.7)
axes[1].set_title('Trajectory by Magnetic Field Magnitude', fontsize=12, fontweight='bold')
plt.colorbar(sc1, ax=axes[1], label='|B|')
for ax in axes: ax.set_xlabel('Longitude'); ax.set_ylabel('Latitude')
plt.tight_layout()
plt.show()
```

```python
# Plot 17: RGB composite - trajectory colored by normalized R,G,B
fig, ax = plt.subplots(figsize=(14, 8))
# Normalize R,G,B to 0-1 for RGB coloring
r = (df['R'] - df['R'].min()) / (df['R'].max() - df['R'].min() + 1e-6)
g = (df['G'] - df['G'].min()) / (df['G'].max() - df['G'].min() + 1e-6)
b = (df['B'] - df['B'].min()) / (df['B'].max() - df['B'].min() + 1e-6)
colors = np.stack([r, g, b], axis=1)
ax.scatter(df['Longitude'], df['Latitude'], c=colors, s=3, alpha=0.8)
ax.set_xlabel('Longitude'); ax.set_ylabel('Latitude')
ax.set_title('ISS Trajectory Colored by RGB (Ambient Light)', fontsize=14, fontweight='bold')
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Plot 18: Time-of-day / elapsed time analysis (sample index as proxy for time)
# Segment data into chunks and compare sensor stats
n_chunks = 20
df['Chunk'] = pd.cut(np.arange(len(df)), bins=n_chunks, labels=False)

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
chunk_means = df.groupby('Chunk')[['Temperature', 'Humidity', 'Pressure', 'Compass']].mean()

for ax, col in zip(axes.flatten(), ['Temperature', 'Humidity', 'Pressure', 'Compass']):
    ax.bar(chunk_means.index, chunk_means[col], color='steelblue', alpha=0.8)
    ax.set_xlabel('Time segment')
    ax.set_ylabel(col)
    ax.set_title(f'{col} by Time Segment (mean)', fontsize=11, fontweight='bold')
    ax.grid(True, alpha=0.3, axis='y')

plt.suptitle('Sensor Averages by Time Segment', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
# Plot 19: Violin plots for Temperature, Humidity, Pressure (by Motion = 0 vs 1 if varied)
if df['Motion'].nunique() > 1:
    fig, axes = plt.subplots(1, 3, figsize=(15, 5))
    for ax, col in zip(axes, ['Temperature', 'Humidity', 'Pressure']):
        sns.violinplot(data=df, x='Motion', y=col, ax=ax)
        ax.set_title(col)
    plt.suptitle('Environmental Sensors by Motion', fontsize=14, fontweight='bold', y=1.02)
else:
    # If Motion is constant, show violin for Pitch, Roll, Yaw
    fig, axes = plt.subplots(1, 3, figsize=(15, 5))
    for ax, col in zip(axes, ['Pitch', 'Roll', 'Yaw']):
        sns.violinplot(y=df[col], ax=ax)
        ax.set_title(col)
    plt.suptitle('Orientation Sensors Distribution', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
# Plot 20: Multi-panel - all environmental + magnetic in one figure (zoomed sample)
sample = df.iloc[::50]  # Every 50th row for clarity

fig, axes = plt.subplots(4, 2, figsize=(16, 14), sharex=True)
cols_grid = [['Temperature', 'Humidity'], ['Pressure', 'Compass'], ['MagX', 'MagY'], ['MagZ', 'Mag_Magnitude']]
for i, row_cols in enumerate(cols_grid):
    for j, col in enumerate(row_cols):
        if col in sample.columns:
            axes[i, j].plot(sample['Date/Time'], sample[col], linewidth=1, alpha=0.9)
        axes[i, j].set_ylabel(col)
        axes[i, j].set_title(col, fontsize=11, fontweight='bold')
        axes[i, j].grid(True, alpha=0.3)

axes[-1, 0].set_xlabel('Time'); axes[-1, 1].set_xlabel('Time')
plt.suptitle('Sensor Time Series (Sampled Every 50th Point)', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
# Plot 21: Scatter - Temperature vs Pressure (colored by Humidity)
sc = plt.scatter(df['Temperature'], df['Pressure'], c=df['Humidity'], cmap='viridis', alpha=0.3, s=10)
plt.colorbar(sc, label='Humidity (%)')
plt.xlabel('Temperature (°C)'); plt.ylabel('Pressure (hPa)')
plt.title('Temperature vs Pressure (colored by Humidity)', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Plot 22: Latitude & Longitude distributions
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].hist(df['Latitude'], bins=60, edgecolor='black', alpha=0.7)
axes[0].set_xlabel('Latitude'); axes[0].set_ylabel('Count')
axes[0].set_title('Latitude Distribution', fontsize=12, fontweight='bold')
axes[1].hist(df['Longitude'], bins=60, edgecolor='black', alpha=0.7, color='green')
axes[1].set_xlabel('Longitude'); axes[1].set_ylabel('Count')
axes[1].set_title('Longitude Distribution', fontsize=12, fontweight='bold')
plt.suptitle('ISS Position Distributions', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```
